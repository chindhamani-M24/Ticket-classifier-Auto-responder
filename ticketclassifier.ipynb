```python
# ==========================================
# Interactive Customer Support Ticket Classifier
# Google Colab Copy-Paste Ready Script
# ==========================================

import subprocess
import sys
import pandas as pd

def install_package(package):
    try:
        __import__(package)
    except ImportError:
        subprocess.check_call([sys.executable, "-m", "pip", "install", package])

print("--- Checking and installing required libraries ---")
install_package("pandas")
install_package("scikit-learn")

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# 1. Create Training Dataset
data = {
    'ticket_text': [
        "I was charged twice for my subscription this month, please refund me.",
        "My payment failed at checkout multiple times, help!",
        "I can't log into my account, it says incorrect password even after resetting.",
        "How do I change my email address in account settings?",
        "The app crashes every time I try to upload a profile picture.",
        "Page loads extremely slowly and gives a 504 gateway timeout error.",
        "Where is my order? Tracking hasn't updated in 5 days.",
        "Can I change the shipping address for my recent purchase?",
        "I want to cancel my subscription and close my account permanently.",
        "How does your pricing tier work for teams?"
    ],
    'category': [
        'Billing', 'Billing',
        'Account', 'Account',
        'Technical', 'Technical',
        'Shipping', 'Shipping',
        'Account', 'Billing'
    ],
    'urgency': [
        'High', 'High',
        'High', 'Low',
        'Medium', 'High',
        'Medium', 'Medium',
        'Medium', 'Low'
    ]
}

df = pd.DataFrame(data)

print("\n--- Training Machine Learning Models ---")
# Train Category Classifier Pipeline
intent_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('clf', LogisticRegression())
])
intent_pipeline.fit(df['ticket_text'], df['category'])

# Train Urgency Classifier Pipeline
urgency_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('clf', LogisticRegression())
])
urgency_pipeline.fit(df['ticket_text'], df['urgency'])

# 2. Automated Response Templates
response_templates = {
    'Billing': "Hello! We apologize for the billing issue. Our finance team is reviewing your transaction and will process any eligible refunds within 24-48 business hours.",
    'Account': "Hello! We understand account access is critical. Please follow the secure recovery link sent to your email address to reset your credentials safely.",
    'Technical': "Hello! Thank you for reporting this technical bug. Our engineering team has been notified and is actively investigating the root cause.",
    'Shipping': "Hello! We are looking into your shipment status with our courier partner. You will receive an updated tracking link shortly."
}

# 3. Interactive Loop for Colab
print("\n" + "="*50)
print("🤖 INTERACTIVE SUPPORT TICKET ASSISTANT READY")
print("Type a customer message below and press Enter.")
print("Type 'exit' or 'quit' to stop.")
print("="*50 + "\n")

while True:
    try:
        ticket = input("💬 Enter customer ticket message: ").strip()
    except EOFError:
        break

    if ticket.lower() in ['exit', 'quit']:
        print("\nExiting interactive assistant. Have a great day!")
        break

    if not ticket:
        print("⚠️ Please enter a valid message.\n")
        continue

    predicted_category = intent_pipeline.predict([ticket])[0]
    predicted_urgency = urgency_pipeline.predict([ticket])[0]
    auto_response = response_templates.get(predicted_category, "Thank you for reaching out. We will get back to you soon.")

    print(f"\n[AI Analysis Result]")
    print(f"👉 Predicted Intent : {predicted_category}")
    print(f"👉 Urgency Level   : {predicted_urgency}")
    print(f"👉 Auto-Response   : {auto_response}")
    print("-" * 50 + "\n")
```
